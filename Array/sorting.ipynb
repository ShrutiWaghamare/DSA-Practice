```python
# selection sort
arr = [64, 25, 12, 22, 11]
n = len(arr)

for i in range(len(arr)):
    min_ind = i
    for j in range(i+1, len(arr)):
        if arr[j] < arr[min_ind]:
            min_ind = j 
    arr[i], arr[min_ind] = arr[min_ind], arr[i]
print(arr)
```

Example

Array:

[64, 25, 12, 22, 11]
Step 1: Divide
[64, 25, 12, 22, 11]

→ [64, 25]     [12, 22, 11]

→ [64] [25]    [12] [22, 11]

→ [22] [11]
Step 2: Merge (Sort while merging)
Merge [64] and [25]
[25, 64]
Merge [22] and [11]
[11, 22]
Merge [12] and [11, 22]
[11, 12, 22]
Final Merge
[25, 64] + [11, 12, 22]

Step-by-step:

11 → 12 → 22 → 25 → 64

Final:

[11, 12, 22, 25, 64]

```python
def merge_sort(arr):
    if len(arr) > 1:
        mid = len(arr)//2

        left = arr[:mid]
        right = arr[mid:]

        merge_sort(left)
        merge_sort(right)

        i = j = k = 0

        while i < len(left) and j < len(right):
            if left[i] < right[j]:
                arr[k] = left[i]
                i += 1
                k += 1
            else:
                arr[k] = right[j]
                j += 1
                k += 1

        while i < len(left):
            arr[k] = left[i]
            i += 1
            k += 1
        
        while j < len(right):
            arr[k] = right[j]
            j += 1
            k += 1
    return arr

arr1 = [64, 25, 12, 22, 11]
print(merge_sort(arr1))
```

Question: Task Scheduling

You are given N tasks.
Each task has:

Priority (smaller value = higher priority)
Time required to complete it

```python
n = int(input())
tasks = []
for _ in range(n):
    p, t = map(int, input().split())
    tasks.append([p, t])
    
for i in range(n):
    min_index = i
    for j in range(i+1, n):
        if tasks[j][0] < tasks[min_index][0]:
            min_index = j
        elif tasks[j][0] == tasks[min_index][0]:
            if tasks[j][1] < tasks[min_index][1]:
                min_index = j
    tasks[i], tasks[min_index] = tasks[min_index], tasks[i]

for task in tasks:
    print(task[0], task[1], end = ",")
    
    
```

```python
# Quick sort
def quick(arr):
    if len(arr) < 1:
        return arr
    
    left = []
    right = []
    pivot = arr[-1]
    
    for i in range(len(arr)-1):
        if arr[i] < pivot:
            left.append(arr[i])
        else:
            right.append(arr[i])
    
    return quick(left) + [pivot] + quick(right)

arr = list(map(int, input().split()))
print(quick(arr))
```

```python
# Linear Search
def linear(arr, target):
    for i in range(len(arr)):
        if arr[i] == target:
            return {target:i}
    return -1

arr = list(map(int, input().split()))
print(linear(arr, 1))
```

```python
# Binary Search
def binary(arr, target):
    low = 0
    high = len(arr)-1
    while low <= high:
        mid = (low + high) // 2
        if arr[mid] == target:
            return mid
        elif arr[mid] < target:
            low = mid + 1
        else:
            high = mid - 1
    return -1

arr = list(map(int, input().split()))
print(binary(arr, 5))
```
